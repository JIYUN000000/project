```python
import numpy as np
from sympy import *

# 행렬 A의 첨가행렬을 기약 행 사다리꼴로 변환
# np.hstack(A, b) : 행렬 A에 벡터 b를 붙여 첨가행렬 만들기
# Matrix(): 행렬을 의미
A = np.array([[1,3,2],[2,2,0],[-3,1,1]])
b = np.array([[2,0,-2]])

augmented_A = Matrix(np.hstack((A, b.T)))
print("행렬 A의 첨가행렬 = \n", np.array(augmented_A))
print("\n")
print("행렬 A의 첨가행렬의 기약 행 사다리꼴 = \n", np.array(augmented_A.rref()[0]))
```

```python
import numpy as np

# 행렬 A의 역행렬 계산
# np.linalg.matrix_power(A,n): 행렬 A의 n제곱
# np.linalg.matrix_power(A,-1): 행렬 A의 역행렬
A = np.array([[2,3],[5,7]])
A_inv = np.linalg.matrix_power(A, -1)
print("Inverse matrix of A = \n", A_inv)
```

```python
import numpy as np

# 행렬 A의 역행렬 계산
# np.linalg.inv(A): 행렬 A의 역행렬
A = np.array([[1,3,2],[2,2,0],[-3,1,1]])
A_inv = np.linalg.inv(A)
print("Inverse matrix of A = \n", A_inv)
```

```python
import numpy as np

# 행렬 A의 전치행렬 구하기
# A.T: 행렬 A의 전치행렬
A = np.array([[-1,3],[3,7]])
A_T = A.T
print("Transposed matrix of A = \n", A_T)
print("\n")

# 행렬 B의 전치행렬 구하기
B = np.array([[1,3,5],[7,2,9],[10,11,3]])
B_T = B.T
print("Transposed matrix of B = \n", B_T)
print("\n")
```

```python
import numpy as np

A = np.array([[1,2,0],[3,2,1],[2,-1,1]])
B = np.array([[1,-2,3],[1,4,2],[-1,0,7]])

# 행렬 AB의 계산과 행렬 AB의 대각합 계산
# np.trace(A): 행렬 A의 대각합
AB = np.matmul(A,B)
tr_AB = np.trace(AB)
print("AB =\n", AB)
print("\n")
print("Trace of AB = ", tr_AB)
print("\n")

# 행렬 BA의 계산과 행렬 BA의 대각합 계산
BA = np.matmul(B,A)
tr_BA = np.trace(BA)
print("BA =\n", BA)
print("\n")
print("Trace of BA = ", tr_BA)
print("\n")

print("행렬 AB와 행렬 BA는 같은가?", np.array_equal(AB, BA))
print("tr(AB)와 tr(BA)는 같은가?", tr_AB == tr_BA)
```
