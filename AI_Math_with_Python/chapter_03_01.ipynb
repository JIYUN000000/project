```python
import numpy as np
# 노름을 구하는 함수
# np.sqrt(q): q^(1/2) 계산
# np.sum(): () 안에 있는 숫자의 합
def cal_norm(vec1):
  norm_vec = np.sqrt(np.sum(vec1*vec1))
  return norm_vec

# x의 노름 계산
x = np.array([1,2,0,2])
print("Norm of x = ", cal_norm(x))
```

```python
import numpy as np

# 내적을 구하는 함수
def cal_innerproduct(vec1, vec2):
  innerprod_vecs = np.sum(vec1*vec2)
  return innerprod_vecs

# x와 y의 내적 계산
x = np.array([1,2,3,-1])
y = np.array([0,1,2,1])
print("Inner product of x and y = ", cal_innerproduct(x,y))
print("\n")

# y와 z의 내적 계산
y = np.array([0,1,2,1])
z = np.array([2023,1,-1,1])
print("Inner product of y and z = ", cal_innerproduct(y,z))
print("\n")
```

```python
import numpy as np

# 내적을 구하는 함수
# np.cos(): ()의 코사인 값, np.pi: π
def cal_innerproduct_v2(norm_vec1, norm_vec2, angle):
  denominator = 180/angle
  innerprod_vecs = norm_vec1 * norm_vec2 * np.cos(np.pi/denominator)
  return innerprod_vecs

# x와 y의 내적 계산
norm_x = 2
norm_y = 3
angle = 30
print("Inner product of x and y = ", cal_innerproduct_v2(norm_x, norm_y, angle))
```

```python
import numpy as np

# 코시-슈바르츠 부등식을 확인하는 함수
# np.abs(): ()의 절댓값
def check_Cauchy_Schwarz_inequality(vec1, vec2):
  innerprod_vecs = np.sum(vec1* vec2)
  norm_vec1 = np.sqrt(np.sum(vec1*vec2))
  norm_vec2 = np.sqrt(np.sum(vec2*vec2))
  check_CS_inequal = np.abs(innerprod_vecs) <= norm_vec1 * norm_vec2
  return check_CS_inequal

# x와 y로 확인
x = np.array([1,-1,3])
y = np.array([1,1,3])
print("Cauchy_Schwarz_inequality is", check_Cauchy_Schwarz_inequality(x,y))
```

```python
import numpy as np

# 직교성을 확인하는 함수
def check_Orthogonal(vec1, vec2):
  innerprod_vecs = np.sum(vec1*vec2)
  check_Orthogonal = innerprod_vecs == 0
  return check_Orthogonal

# x와 y로 확인
x = np.array([2,-5,1])
y = np.array([2,1,1])
print("Orthogonal of x and y is", check_Orthogonal(x,y))
print("\n")

# y와 z로 확인
y = np.array([2,1,1])
z = np.array([4,1,-3])
print("Orthogonal of y and z is", check_Orthogonal(y,z))
print("\n")

# x와 z로 확인
x = np.array([2,-5,1])
z = np.array([4,1,-3])
print("Orthogonal of x and z is", check_Orthogonal(x,z))
print("\n")
```

```python
import numpy as np

# 정사영을 구하는 함수
def proj_a_b(vec_1, vec_2):
  p = (np.dot(vec_1,vec_2)/np.dot(vec_2, vec_2))*vec_2
  return p

x = np.array([1,1])
y = np.array([2,0])

print("Projection of x onto y = ", proj_a_b(x,y))
```

```python
import numpy as np

# 해밍 거리를 구하는 함수
# len(vec): vec의 성분 개수
def cal_hamming_dist(vec1, vec2):
  Num_vec = len(vec1)
  hamming_dist = 0
  for i in range(Num_vec):
    if vec1[i] != vec2[i]:
      hamming_dist = hamming_dist + 1
  return hamming_dist

# v와 w 사이의 해밍 거리 계산
v = np.array([0,1,1,0,1])
w = np.array([1,1,0,1,0])
print("Hamming distance = ", cal_hamming_dist(v,w))
```

```python
import numpy as np

# 맨하튼 거리를 구하는 함수
def cal_Manhattan_dist(vec1, vec2):
  Manhattan_dist = np.sum(np.abs(vec1-vec2))
  return Manhattan_dist

# A = (1,1), B = (3,3)이라 할 때, 두 점 A, B 사이의 맨하튼 거리 계산
A = np.array([1,1])
B = np.array([3,3])
print("Manhattan distance = ", cal_hamming_dist(A,B))
```

```python
from sympy import *
import numpy as np

# 문제 (a)
# symbols('x'): x를 미지수로 선언
# diff(f,x): 함수 f를 x로 미분
x, y, z = symbols('x y z')
f = x*y*z

df_dx = diff(f, x)
df_dy = diff(f, y)
df_dz = diff(f, z)

gradient_f_1 = np.array([[df_dx], [df_dy], [df_dz]])
print("(a) gradient of f = \n", gradient_f_1)
print("\n")

# 문제 (b)
x, y, z = symbols('x y z')
f = x**2*y**3 + z**2 + exp(x*y)

df_dx = diff(f,x)
df_dy = diff(f,y)
df_dz = diff(f,z)

gradient_f_1 = np.array([[df_dx], [df_dy], [df_dz]])
print("(b) gradient of f = \n", gradient_f_1)
```

```python
from sympy import *
import numpy as np

x, y, z = symbols('x y z')
F_1 = x-y+z**2
F_2 = x*cos(y)
F_3 = y**2*exp(3*z)

# F1, F2, F3를 x로 미분
df1_dx = diff(F_1, x)
df2_dx = diff(F_2, x)
df3_dx = diff(F_3, x)

# F1, F2, F3를 y로 미분
df1_dy = diff(F_1, y)
df2_dy = diff(F_2, y)
df3_dy = diff(F_3, y)

# F1, F2, F3를 z로 미분
df1_dz = diff(F_1, z)
df2_dz = diff(F_2, z)
df3_dz = diff(F_3, z)

Jacobian_F = np.array([[df1_dx,df2_dx,df3_dx],[df1_dy,df2_dy,df3_dy], [df1_dz,df2_dz,df3_dz]])
print("Jacobian_F = \n",Jacobian_F)
```

```python
from sympy import *
import numpy as np

# diff(f,x,n): 함수 f를 x로 n 번 미분
# diff(f,x,y): 함수 f를 x와 y 순으로 미분
x, y, z = symbols('x y z')
f = x**2 - y**2 + z**3

d2f_dx2 = diff(f,x,2)
d2f_dyx = diff(f,y,x)
d2f_dzx = diff(f,z,x)

d2f_dxy = diff(f,x,y)
d2f_dy2 = diff(f,y,2)
d2f_dzy = diff(f,z,y)

d2f_dxz = diff(f,x,z)
d2f_dyz = diff(f,y,z)
d2f_dz2 = diff(f,z,2)

Hessian_f = np.array([[d2f_dx2,d2f_dyx,d2f_dzx],[d2f_dxy,d2f_dy2,d2f_dzy],[d2f_dxz,d2f_dyz,d2f_dz2]])
print("Hessian_f = \n", Hessian_f)
```

```python
from sympy import *
import numpy as np

# 문제 (a)
x, y = symbols('x y')
f = x**3 + 2*x + y**2

d2f_dx2 = diff(f,x,2)
d2f_dy2 = diff(f,y,2)

Laplacian_f = d2f_dx2 + d2f_dy2
print("(a) Laplacian of f = ", Laplacian_f)
print("\n")

# 문제 (b)
x, y, z = symbols('x y z')
f = x**3*y*z**2 + 2*cos(y) + x*exp(y*z)

d2f_dx2 = diff(f,x,2)
d2f_dy2 = diff(f,y,2)
d2f_dz2 = diff(f,z,2)

Laplacian_f = d2f_dx2 + d2f_dy2 + d2f_dz2
print("(b) Laplacian of f = ", Laplacian_f)
print("\n")
```
